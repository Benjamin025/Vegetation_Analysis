```python
import ee
import json
import datetime
import geemap
import pandas as pd
```

```python
# Create an interactive map centered on your region
Map = geemap.Map(center=[0, 20], zoom=3)
Map.add_basemap("SATELLITE")
Map.add_draw_control()  # Allow drawing polygons, rectangles, etc.

Map
```

```python
# Once you draw an AOI on the map, run this cell to get the geometry
aoi = Map.user_roi

if aoi is None:
    raise Exception("Please draw an AOI on the map before running this cell.")

Map.centerObject(aoi)
aoi
```

```python
# Select the MODIS collection
modis = ee.ImageCollection("MODIS/061/MOD13Q1").select(["NDVI", "EVI"])

# Choose year range or custom start-end
start_date = "2024-01-01"
end_date = "2024-12-31"

modis_year = modis.filterBounds(aoi).filterDate(start_date, end_date)

# MODIS scale factor correction (0.0001)
def scale_modis(img):
    return img.multiply(0.0001).copyProperties(img, img.propertyNames())

modis_scaled = modis_year.map(scale_modis)
print("Number of images:", modis_scaled.size().getInfo())
```

```python
# Load MODIS NDVI and EVI
modis = (ee.ImageCollection("MODIS/006/MOD13Q1")
         .select(["NDVI", "EVI"])
         .filterDate(f"{year}-01-01", f"{year+1}-01-01")
         .filterBounds(aoi))

# Optional: scale NDVI/EVI to real values
def scale_modis(img):
    return img.multiply(0.0001).copyProperties(img, img.propertyNames())

modis = modis.map(scale_modis)

# Compute NDVI min and max over the year for VCI baseline
ndvi_min = modis.select("NDVI").reduce(ee.Reducer.min()).rename(["NDVI_min"])
ndvi_max = modis.select("NDVI").reduce(ee.Reducer.max()).rename(["NDVI_max"])

# Add VCI band to each image
def add_vci(img):
    vci = img.expression(
        "((ndvi - ndvi_min) / (ndvi_max - ndvi_min)) * 100",
        {
            "ndvi": img.select("NDVI"),
            "ndvi_min": ndvi_min.select("NDVI_min"),
            "ndvi_max": ndvi_max.select("NDVI_max")
        }
    ).rename("VCI")
    return img.addBands(vci)

modis_indices = modis.map(add_vci)
```

```python
def monthly_composites(collection, year):
    results = []
    for m in range(1, 13):
        start = ee.Date.fromYMD(year, m, 1)
        end = start.advance(1, "month")

        monthly = collection.filterDate(start, end).mean().clip(aoi)
        stat = monthly.reduceRegion(
            reducer=ee.Reducer.mean(),
            geometry=aoi,
            scale=250,
            maxPixels=1e13
        )
        results.append({
            "Month": m,
            "NDVI": stat.get("NDVI"),
            "EVI": stat.get("EVI"),
            "VCI": stat.get("VCI")
        })
    return results


year = 2024
monthly_stats = monthly_composites(modis_indices, year)
```

```python
def ee_stat_to_dict(stat):
    """Convert Earth Engine dictionary values to client-side numbers."""
    return {
        "Month": stat["Month"],
        "NDVI": stat["NDVI"].getInfo() if isinstance(stat["NDVI"], ee.ComputedObject) else stat["NDVI"],
        "EVI": stat["EVI"].getInfo() if isinstance(stat["EVI"], ee.ComputedObject) else stat["EVI"],
        "VCI": stat["VCI"].getInfo() if isinstance(stat["VCI"], ee.ComputedObject) else stat["VCI"],
    }

# Convert all months
monthly_stats_client = [ee_stat_to_dict(s) for s in monthly_stats]

import pandas as pd
df = pd.DataFrame(monthly_stats_client)
df
```
